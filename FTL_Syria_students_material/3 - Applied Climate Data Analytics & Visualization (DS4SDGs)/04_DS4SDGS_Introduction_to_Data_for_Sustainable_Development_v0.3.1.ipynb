## 1.1. Definition of SDGs

The [Sustainable Development Goals (SDGs)](https://sdgs.un.org/goals) were established by the United Nations (UN) in 2015 as a global call to action to end poverty, protect the planet, and ensure peace and prosperity for all by 2030.

The 17 SDGs are interconnected and focus on achieving a balance between social, economic, and environmental sustainability. Developed nations play a crucial role in supporting those most disadvantaged in achieving these goals. Achieving the SDGs requires collective efforts, drawing on creativity, technological innovation, and financial resources from all sectors of society.

<br>

Below is a list of the 17 SDGs as adopted by the UN.


**SDG 1. No Poverty**   
Understand the spending patterns on mobile phone services as a proxy indicator of income levels

**SDG 2. Zero Hunger**   
Monitor food prices listed online or crowdsourcing to track food security changes

**SDG 3. Good Health and Well-Being**   
Predict the spread of infectious diseases by mapping the movement of anonymized mobile phone users

**SDG 4. Quality Education**   
Curriculum and course recommendation optimized for an individual's learning need

**SDG 5. Gender Equality**   
Analysis of financial transaction to cluster spending patterns and different levels of economic resilience based on a gender

**SDG 6. Clean Water and Sanitation**   
Internet of Things (IoT) sensors connected to water pumps to track the hygiene level of an individual well

**SDG 7. Affordable and Clean Energy**   
Smart metering apporach to increase or restrict the flow of electricity, gas, or water, which could reduce ensure adequate supply for all at peak periods

**SDG 8. Decent Work and Economic Growth**   
Analysis with global postal traffic to measure the economic growth, remittances, trade, and GDP

**SDG 9. Industry, Innovation and Infrastructure**   
Locational data generated by mobile devices to optimize public transportation routes

**SDG 10. Reduced Inequality**   
Speech-to-text analytics on a local radio content to identify concerns and policy responses from the public

**SDG 11. Sustainable Cities and Communities**   
Remote sensing with satellite imagery to determine whether sufficient green spaces are available per population concentration

**SDG 12. Responsible Consumption and Production**   
Analytics on the online purchase data to identify the pace of transition from non-energy efficient to energy-efficient products.

**SDG 13. Climate Action**    
Deforestation tracker based on combined satellite imagery, crowd-sourced witness accounts, and open data

**SDG 14. Life Below Water**   
Location tracker for the maritime vessels to detect illegal or unregulated fishing activities

**SDG 15. Life on Land**   
Social media monitoring to support disaster management with real-time information on victim location, effects and strength of a forest fire

**SDG 16: Peace, Justice and Strong Institutions**   
Sentiment analysis of social media to reveal public opinion on effective governance, public service delivery or human rights.

**SDG 17: Partnerships to Achieve the Goal**   
Private-Public partnership to enable integration of national statistics along with mobile and internet data for synergy.

## 2.3. Data Commons

[Data Commons](https://datacommons.org/), developed by Google, is a comprehensive platform that consolidates public data from various sources, such as government agencies and international organizations, making it accessible and easy to use. Its primary goal is to simplify data-driven decision-making and provide insights into societal challenges.

**Key Features:**

* **Unified Knowledge Graph:**
Data Commons integrates diverse datasets into a single knowledge graph, connecting related information. This eliminates the need for extensive data cleaning, allowing users to access specific insights effortlessly.

* **Accessibility:**
It is open to all users, providing tools for researchers, educators, and policymakers, along with APIs for developers to build applications.

* **Use Cases:**
Designed to support education, research, and decision-making by offering a reliable and centralized repository of data.


**How It Works:**

* **Nodes and Edges:** Entities, such as locations or statistics, are represented as nodes, while relationships between these entities are denoted as edges.

  * [Santa Clara County](https://browser.datacommons.org/kg?dcid=geoId/06085) is contained in the [State of California](https://browser.datacommons.org/kg?dcid=geoId/06)
  * The latitude of [Berkeley, CA](https://browser.datacommons.org/kg?dcid=geoId/0606000) is 37.8703
  * [The population of Maryland](https://browser.datacommons.org/kg?dcid=dc/p/psjx4xy30nws1) was [6,003,435 in 2018](https://browser.datacommons.org/kg?dcid=dc%2Fo%2Fx1tlfg4ll9yr9).

* **Schema.org Integration:** Data Commons adheres to the [Schema.org data model](https://schema.org/docs/datamodel.html), leveraging standardized schemas to ensure consistency in data representation, including types and properties.

### 2.3.1. Accessing Data Commons Data

**Data Commons Terms to Understand**

* **DCIDs (Data Commons Identifiers):**
Each entity in the Data Commons knowledge graph is assigned a unique identifier called a dcid. This identifier allows users to search for and query specific information about an entity, enabling precise and efficient data retrieval.

* **StatVars (Statistical Variables):**
A Statistical Variable represents a measurable metric, statistic, or characteristic observed at a particular place and time. Examples include:

  * Median income of individuals aged 16 and older.
  * Number of female high school graduates aged 18 to 24.
  * Unemployment rate.
  * Percentage of individuals diagnosed with diabetes.

  StatVars are also represented as nodes within the knowledge graph and can be explored directly using the graph browser, offering a structured way to access diverse data points.

## 2.5. United Nations Statistics Division (UNSD) SDG API


The [UNSD SDG API](https://unstats.un.org/sdgs/UNSDGAPIV5/swagger/index.html), provided by the United Nations Statistics Division, offers access to official Sustainable Development Goals (SDG) data reported by custodian agencies. This API allows you to retrieve data on various SDG indicators, geographical areas, time periods, and more.

**Key API Endpoints**

The UNSD SDG API includes several categories of endpoints, each providing specific types of data:

- **Archive Data:** Historical SDG data.
- **Compare Trends:** Data for analyzing trends across indicators or regions.
- **Data Availability:** Information on the availability of SDG data.
- **GeoArea:** Geographical information related to SDGs.
- **Global Database:** Comprehensive global SDG datasets.
- **Goal:** Data related to specific SDGs.
- **Indicator:** Detailed data for each goal's indicators.
- **Series:** Time-series data for various SDG metrics.
- **Target:** Data on specific SDG targets.
- **User:** Information for managing API access and usage.

Each endpoint corresponds to its category. For example, the Goal endpoint provides data related to specific SDGs, while the Indicator endpoint delivers detailed insights for individual indicators within each goal.






### 2.5.1. Accessing SDG Data using the API

The base URL for the **UNSD SDG API** is:
https://unstats.un.org/sdgs/UNSDGAPIV5/swagger/index.html

````python
#imprt relevant library
import requests
import pandas as pd
````

**1. Get a list of all SDG Goals.**

````python
base_url = "https://unstats.un.org/sdgs/UNSDGAPIV5"


#get list of SDGs
response = requests.get(f"{base_url}/v1/sdg/Goal/List")

#check returned request
if response.status_code == 200:
    goals = response.json()
else:
    print(f"Request failed with status code {response.status_code}: {response.text}")

print(goals)
````

**2. Get details about a specific Goal (e.g., Goal 1).**

If you are interested in more details for Goal 1 for example, you can use the endpoint `GET /v1/sdg/Goal/{goalCode}/Target/List` to get a list of all targets available for this goal.


````python
goal_code = "1"
response = requests.get(f"{base_url}/v1/sdg/Goal/{goal_code}/Target/List")

#check returned request
if response.status_code == 200:
    specific_goal = response.json()
else:
    print(f"Request failed with status code {response.status_code}: {response.text}")

print(specific_goal)
````

### 3.4.1. Geospatial Interactive Visualization

This section focuses on plotting health facilities on a map using their latitude and longitude coordinates. The map will also differentiate between facility owners by displaying markers in distinct colors for each ownership type.

````python
# install and import folium to create interactive maps,
!pip install folium

import folium
````

````python
def get_color(ownership):

    """
    Returns a color based on the ownership type.
    Defaults to 'black' if ownership type is not found.
    """

    colors = {
        'CHAG': 'blue',
        'Private': 'green',
        'private': 'lightgreen',
        'Government': 'red',
        'government': 'darkred',
        'Quasi-Government': 'purple',
        'Islamic': 'orange',
        'Muslim': 'darkorange',
        'Maternity Home': 'pink',
        'Clinic': 'lightblue',
        'NGO': 'gray',
        'Mission': 'brown'
            }
    return colors.get(ownership, 'black')  # Default to 'black' if ownership type not found


# Drop rows with any NaN values
facilities_data = facilities.dropna()

# Create a map centered around Ghana
ghana_map = folium.Map(location=[7.9465, -1.0232], zoom_start=6)

# Iterate through the facilities data and add small circle markers
for index, row in facilities_data.iterrows():
    # Customize the popup text
    popup_text = f"Facility: {row['FacilityName']}<br> Ownership: {row['Ownership']}"
    color = get_color(row['Ownership'])

    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=3, # You can adjust the size of the dot here
        color=color,
        fill=True,
        fill_opacity=1,
        popup=popup_text
    ).add_to(ghana_map)

# Show the map
ghana_map
````
