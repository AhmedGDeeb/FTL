### 2.2.2. **Raster Data**

Pixels or grid cells make up raster data. They are often square and spaced regularly. Rasters, however, can also be rectangular. Each pixel is associated with a value (**Figure 5**). Continuous rasters such as temperature or elevation, have values that gradually change. Discrete rasters, however, assign each pixel to a specific class. For instance, land cover classes are represented to a set of values. Raster data can come in a variety of types. The ones that are commonly used are tiff, xyz, .nc (also for vector), jpg, jp2 and more.
They are mostly images with some geographical context meta-data attached to it (**Figure 6**). GIS uses these meta files to find the raster files on a map.\
(Source: <a href=https://gisgeography.com/what-is-geodata-geospatial-data/>GIS Geography</a>)

## 2.5. Geographic and Projected Coordinate Systems

In working with geospatial data, understanding coordinate systems is crucial to referencing locations across the globe. In this section, we will discuss the systems and method to align geospatial data on the Earth's surface.

<p align="center">
<center><img src="https://mapscaping.com/wp-content/uploads/2024/09/image-592.png" width="500"/></center>
</p><center><b>Figure 12: </b>Map projection types <i>(<a href=https://mapscaping.com/>source</a>)</i></center>

<p></p>

**1. Geographic Coordinate Systems (GCS)**

A Geographic Coordinate System is used to locate features on the Earth's surface. It represents locations using a grid defined by latitude and longitude, which are measured in degrees.

Latitude: Measures north-south position from the equator (0°).\
Range: -90° (South Pole) to +90° (North Pole).\
Longitude: Measures east-west position from the Prime Meridian (0°).\
Range: -180° to +180°.

A GCS also includes:\
Datum: A mathematical model of the Earth's shape (e.g., WGS84, NAD83).\
Units: Typically degrees.\
*Example:*
`A point in New York: (40.7128° N, 74.0060° W).`

**2. Projected Coordinate System (PCS)**

The Earth's surface is a 3D curved shape (an ellipsoid), but maps are flat. A projection is the method used to transform the 3D surface into a 2D plane (**Figure 12**). This projection in defined in a projected coordinate system.

*Why Use Projections?* \
Geographic coordinates are great for global positioning but not ideal for measuring distances, areas, or angles directly.

Projections allow us to:
- Display the Earth’s surface on flat maps.
- Perform spatial analyses in Cartesian (x, y) coordinates with metric units (e.g., meters).

**3. Common Projections in GIS**
- *Web Mercator (EPSG:3857):* Common for online maps (e.g., Google Maps, OpenStreetMap). This projection distorts area but preserves shapes.
- *Universal Transverse Mercator (UTM):* Divides the Earth into 60 zones, each with its own projection. It is useful for small-scale, high-accuracy mapping.
- *Albers Equal-Area:* Maintains area accuracy. It is often used for regional or national maps.
- *Lambert Conformal Conic:* Balances shape and area distortion. Common in aviation and meteorology.

    - For instance, SAR (Synthetic Aperture Radar), like Sentinel-1, or advanced LiDAR (Light Detection and Ranging) missions can penetrate forest canopies, offering insights that would otherwise remain obscured.

### a) Unsupervised Classification

**Unsupervised Classification** refers to the process of categorizing pixels with similar characteristics based on software analysis of an image. This method does not require the user to provide sample classes. Instead, the computer uses algorithms to analyze pixel data and group them into distinct classes.

Although the user can choose the algorithm and specify the number of output classes, their involvement in the classification process is minimal. However, users must have knowledge of the area being classified to associate pixel groupings with real-world features, such as wetlands, urban areas, or forests.

In this method, reflectance data is used without the need for training data. While this approach may seem unconventional, it is particularly useful when:
- There is limited prior information about the study area.
- There is general knowledge of land cover class distributions but no specific ground data available.

The algorithm identifies clusters of pixels with similar spectral characteristics, making it easier to analyze and interpret the image.

**K-means clustering** is one of such unsupervised image classification algorithms. This algorithm groups pixels into clusters based on the statistical similarity of their spectral values (see **Figure 21**).

## **6.2. Machine Learning Alghoritms**

## **a) Random Forest (RF)**

Random Forest is an ensemble learning method that builds multiple decision trees during training and outputs the mode (classification) or mean (regression) prediction of the individual trees. It is known for its robustness and ability to handle complex relationships in large datasets.

### Advantages in Eco-tourism Suitability Analysis
- Handles both categorical and continuous data effectively.
- Resistant to overfitting, making it ideal for spatial data with many variables.- Provides feature importance, which is useful for understanding which factors (e.g., proximity to rivers, land cover types) most influence suitability.
- Can model non-linear relationships and interactions between variables without explicit specification.

### Applications
In ecotourism suitability analysis, RF can classify areas based on multiple factors (e.g., ecological, geological, and human factors) and generate suitability maps by learning from training data.

## **b) Support Vector Machine (SVM)**

SVM is a supervised learning algorithm that is particularly effective for classification tasks. It works by finding the hyperplane that best separates data into different classes. For regression tasks, it uses a similar concept but with continuous output.

### Advantages in Eco-tourism Suitability Analysis
- Works well with both linear and non-linear data using kernel functions.
- Effective in high-dimensional spaces (e.g., when there are many spatial features to consider).
- Suitable for small to medium-sized datasets, making it ideal when you have high-quality training data but fewer samples.
- It is particularly effective for binary classification tasks, such as identifying suitable vs. unsuitable areas.

### Applications
SVM can be used for binary classification in ecotourism suitability, distinguishing between suitable and unsuitable areas for ecotourism development based on the presence or absence of specific features.

## **c) XGBoost**

XGBoost (Extreme Gradient Boosting) is an advanced boosting algorithm based on decision trees. It builds trees sequentially, where each tree corrects the errors of the previous one. XGBoost is known for its speed, performance, and efficiency, especially in handling large datasets.

### Advantages in Eco-tourism Suitability Analysis
- High predictive accuracy, often outperforming other algorithms in classification and regression tasks.
- Handles missing values and categorical data natively.
- Works well with large, complex datasets, which are common in environmental and geospatial modeling.
- Includes regularization to prevent overfitting, a significant advantage when working with large sets of input variables.
- Can be easily tuned with hyperparameters to improve model performance.

### Applications
XGBoost is ideal for ecotourism suitability analysis where you have a large number of variables (e.g., climate, land cover, elevation) and need to make accurate predictions on which areas are most suitable for ecotourism activities.

## **d) AdaBoost**

AdaBoost (Adaptive Boosting) is another ensemble method that combines the predictions of multiple weak classifiers to create a strong model. It sequentially applies weak classifiers to the training data, with each classifier focusing more on the misclassified points.

### Advantages in Eco-tourism Suitability Analysis
- Performs well with less complex models and is less likely to overfit compared to other ensemble methods.
- Can handle both regression and classification tasks efficiently.
- Works well when the dataset contains noise or when the model needs to focus on harder-to-classify instances (e.g., ambiguous areas for ecotourism suitability).
- Generally faster than XGBoost in training, although it may not always reach the same level of predictive accuracy in more complex datasets.

### Applications
AdaBoost can be used to classify areas of a landscape as suitable or unsuitable for ecotourism, especially when there is a need for a model that can handle noisy or imbalanced data.

**Compute Model Accuracy, Precision, Recall, F1-score**

To have a better understanding of the performance of the Random Forest Classifier, we can use `sklearn classification report` metric to compute and display the accuracy, precision, recall, and F1-score of the model.

**Overall accuracy** is a metric that measures the proportion of correctly classified instances out of the total instances in a dataset. It provides a general measure of the model's performance but may not be sufficient for imbalanced datasets, where one class dominates. In such cases, additional metrics like precisioin, recall, or F1-score are necessary for a comprehensive model evaluation.

| Metric      | Focuses on               | Penalizes                 | Use Case                                                                    |
|-------------|--------------------------|---------------------------|-----------------------------------------------------------------------------|
| **Precision** | Quality of positive predictions | False positives (FP)      | When false positives are costly or undesirable (e.g., email spam filtering). |
| **Recall**    | Coverage of actual positives  | False negatives (FN)      | When missing positive cases is costly (e.g., detecting cancer).             |
| **F1-Score**  | Balance of precision and recall | Imbalance in precision/recall | When both false positives and false negatives matter equally.               |


````python
# Library for classification report
from sklearn.metrics import classification_report

# Predict the classes using the random forest classifier on the test data
y_pred_rf = rf.predict(X_test)
y_pred_xgb = xgb.predict(X_test_xgb)
y_pred_svm = svm.predict(X_test)
y_pred_ada = ada.predict(X_test)

# Generate the classification report
report_rf = classification_report(y_test, y_pred_rf)
report_xgb = classification_report(y_test_xgb, y_pred_xgb)
report_svm = classification_report(y_test, y_pred_svm)
report_ada = classification_report(y_test, y_pred_ada)

# Print the classification report
print(report_rf)
print(report_xgb)
print(report_svm)
print(report_ada)
````

## 6.11. Plot Results: Barplot of suitability areas from the four models

Despite the fact that map plots make it easier for us to see the spatial distribution, it is still difficult to compare each model statistically.

We can show the results into bar plots for each model. Let us calculate the different in predicted areas for comparison.

````python
# Get resolution (replace with your actual resolution)
resolution = 30  # meters per pixel

# Calculate area per pixel
pixel_area = resolution * resolution  # square meters

# Calculate area for each class in each prediction
rf_class_areas = [np.sum(pred_rf_merge_reshaped == class_val) * pixel_area for class_val in range(3)]
xgb_class_areas = [np.sum(pred_xgb_merge_reshaped == class_val) * pixel_area for class_val in range(3)]
svm_class_areas = [np.sum(pred_svm_merge_reshaped == class_val) * pixel_area for class_val in range(3)]
ada_class_areas = [np.sum(pred_ada_merge_reshaped == class_val) * pixel_area for class_val in range(3)]

# rf_class_areas, xgb_class_areas, etc., will contain the areas for each class (0, 1, 2) for their respective models
````

````python
import matplotlib.pyplot as plt
import numpy as np

# Class names
class_names = ['Non-tourist', 'Potentially Tourist', 'Tourist']

# Model names
model_names = ['Random Forest', 'XGBoost', 'SVM', 'AdaBoost']

# Combine class areas for all models (transposed)
all_class_areas = [rf_class_areas, xgb_class_areas, svm_class_areas, ada_class_areas]
all_class_areas = np.array(all_class_areas).T  # Transpose to have classes as rows

# Set up bar plot
bar_width = 0.2  # Width of each bar
x_positions = np.arange(len(model_names))  # X positions for bars

fig, ax = plt.subplots(figsize=(10, 6))

# Create bars for each class
for i, class_areas in enumerate(all_class_areas):
    ax.bar(x_positions + i * bar_width, class_areas, width=bar_width, label=class_names[i])

# Set plot labels and legend
ax.set_xlabel('Model')
ax.set_ylabel('Area (sq meters)')
ax.set_title('Comparison of Predicted Class Areas by Model')
ax.set_xticks(x_positions + bar_width * (len(class_names) - 1) / 2)
ax.set_xticklabels(model_names)
ax.legend()

plt.show()
````

**Comment:** As can be seen on the bar chart above, SVM predicts most areas as.
