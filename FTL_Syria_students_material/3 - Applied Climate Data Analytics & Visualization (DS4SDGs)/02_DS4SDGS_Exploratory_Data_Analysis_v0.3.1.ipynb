Exploratory Data Analysis (EDA) is the first step in analyzing a dataset and an essential part of preparing data for predictive modeling. ## 1.3. EDA Key Steps for a Data Science Project <a class="anchor" id="one-three"></a>

steps-for-mastering-exploratory-data-analysis.jpg


<b>Figure 2. Steps of Exploratory Data Analysis ([Source](https://www.dasca.org/world-of-data-science/article/a-comprehensive-guide-to-mastering-exploratory-data-analysis))<b>

The EDA process is broad and flexible, with steps tailored to the specific data and questions being addressed. Below is a general summary of common EDA steps that are essential in data science projects:

**1. Examine the Dataset Structure:**

- Review the dataset’s shape by checking the total number of rows and columns, as well as the data types for each column.

**2. Identify Missing and Duplicate Data:**
- Check for null values in the dataset and address them appropriately (e.g., imputation or removal).
- Detect duplicate entries and decide whether they should be removed or retained based on the context.

**3. Analyze Numeric Variables:**

- Visualize the distribution of numeric variables using histograms or density plots to identify patterns, skewness, or outliers.
- Explore relationships between numeric variables through pairwise joint distribution plots or correlation matrices.

**4. Examine Categorical Variables:**

- Plot the frequency distribution of categorical variables using bar charts or pie charts to understand the proportion of each category.

**5. Evaluate Time Series Data (if applicable):**

- Analyze trends and seasonality in time series data by grouping it into daily, monthly, and yearly frequencies.

**6. Check for Outliers:**

- Identify and investigate outliers using boxplots or scatter plots, as they can significantly impact model performance.

**7. Understand Relationships Between Variables:**

- Use techniques like cross-tabulation for categorical data or scatter plots for numeric variables to explore interactions and potential dependencies.

**8. Summarize Key Insights:**

- Create summary statistics (mean, median, standard deviation, etc.) for numeric variables and frequency tables for categorical variables to provide a clear overview of the data.

### 2.1. Introduction to Pandas DataFrame and Series

In the pandas library, a [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html) is described as a "one-dimensional ndarray with axis labels (including time series)."

* You can create a Series from various data types, such as array-like objects (e.g., Python lists or NumPy arrays), iterables, dictionaries, or even scalar values.
* Math operators like +, -, /, and * can be applied to combine two Series objects, regardless of their size.

In contrast, a [DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html) is defined as a "two-dimensional, size-mutable, and potentially heterogeneous tabular data structure."

* A DataFrame can be created from n-dimensional arrays (structured or homogeneous), iterables, or dictionaries.





**Create Pandas Series and DataFrame**

To work with a Pandas Series or DataFrame—or perform any operation using the Pandas library—we first need to import it. This requirement applies to any Python library.

Let's start by importing Pandas and then demonstrate how to create a simple Pandas Series and DataFrame.

````python
#P erform basic operations with Pandas Series

# Addition
print(f"Addition:\n{series_1 + series_2}")

# Subtraction
print(f"\nSubstraction:\n{series_2 - series_1}")

# Multiplication
print(f"\nMultiplication:\n{series_1 * series_2}")

# Division
print(f"\nDivision:\n{series_2 / series_1}")

# Exponent
print(f"\nExponent:\n{series_1 ** series_2}")
````

**Using `loc[]` function of the Dataframe**

The `loc[]` function in a pandas DataFrame allows you to access specific rows and columns. Here's the key difference between `loc[]` and `iloc[]`:

* `loc[]`: Selects rows and columns using labels or names.
* `iloc[]`: Selects rows and columns using integer-based indexing (zero-based position).

One of the main advantages of pandas DataFrames is their ease of use. This becomes evident when working with methods like `loc[]` and `iloc[]`, which are among the most commonly used tools for selecting or filtering rows and columns.

````python
print("Given Dataframe :\n", df)

print("\nIterating over rows using loc function :\n")

# iterate through each row and select
# 'area' and 'year' column respectively.
for i in range(len(df)):
    print(df.loc[i, "area"], df.loc[i, "year"])

    if i%10 == 0:
      break
````

**Using `iterrows()` method of the Dataframe**

Pandas `DataFrame.iterrows()` is used to iterate over DataFrame rows. This returns (index, Series) where the index is an index of the Row and Series is data or content of each row.

````python
print("\nIterating over rows using iterrows() method :\n")

counter = 0

# iterate through each row and select
# 'area' and 'year' column respectively.
for index, row in df.iterrows():

  #print only first 10 rows
    if counter < 10:
        print(row["area"], row["year"])
        counter += 1
    else:
        break
````

**Using `itertuples()` method of the Dataframe**

The `itertuples()` method in a pandas DataFrame is widely used for iterating over rows. It returns an iterator where each row is represented as a tuple.

Key Differences Between `iterrows()` and `itertuples()`:

- Performance: `itertuples()` is faster than `iterrows()`.
- Data Type Preservation: Unlike `iterrows()`, which returns a Series for each row (and does not retain column data types), `itertuples()` preserves the data types of the columns.

When using `itertuples()`, the first element in the tuple is the row's index, followed by the values of each column in the row. This makes it an efficient and convenient option for row-wise iteration.

````python
print("\nIterating over rows using itertuples() method :\n")

counter = 0
# iterate through each row and select
# 'area' and 'year' column respectively.
for row in df.itertuples(index=True, name='Pandas'):
    if counter < 10:
        print(getattr(row, "area"), getattr(row, "year"))
        counter += 1
    else:
        break
````

# 4. Data Cleaning

Data cleaning is an essential part of any data analysis process. It involves preparing and refining the dataset to ensure it is accurate, consistent, and ready for meaningful insights. Below are the key aspects of effective data cleaning:

* Identify missing values and outliers.
* Check for plausibility and errors, such as typos or inconsistencies.
* Identify highly correlated variables or variables with very low variance.
* Spot variables with unusual names or values that might need further inspection.
* Verify variable classes (e.g., characters, factors, integers) to ensure correct data types.
* Rename variables or values, especially for those that are numerous or particularly important.
* Review overall patterns using statistical and numerical summaries.

The Pandas functions discussed earlier are invaluable for understanding your data and implementing appropriate cleaning steps.
<br>

Data cleaning is not just technical. It requires understanding the dataset's context and making thoughtful decisions about how to handle inconsistencies or gaps.
Trade-offs are inevitable. Every choice in data cleaning—whether to remove, impute, or transform data—has implications for subsequent analysis.

## 4.1. Handling Missing Values

Missing data is one of the most frequent issues encountered in datasets. For instance, imagine conducting a customer segmentation analysis where 50% of the records lack address information. This could skew the results, leading to a biased analysis that inaccurately shows no customers in certain regions. This is why identifying and handling missing data is a critical step in the data cleaning process.

Blank space, None, NaN, and NaT are the general types of missing data.

### 4.3.9. df.merge():  


The `pd.merge()` function combines two DataFrames based on common columns or indices. The type of merge is determined by the how parameter.

Merge Types:
- **left:** Includes only the keys from the left DataFrame.
- **right:** Includes only the keys from the right DataFrame.
- **inner:** Includes the intersection of keys from both DataFrames.
- **outer:** Includes the union of keys from both DataFrames.

Each method will be demonstrated using DataFrames with the columns "area", "year", and "life_expectancy".

````python
import pandas as pd

# Create two sample dataframes
df1 = pd.DataFrame({
    'area': ['Area1', 'Area2', 'Area3', 'Area4'],
    'year': [2001, 2002, 2003, 2004],
    'life_expectancy': [75, 78, 82, 85]
})

df2 = pd.DataFrame({
    'area': ['Area3', 'Area4', 'Area5', 'Area6'],
    'year': [2003, 2004, 2005, 2006],
    'life_expectancy': [83, 86, 79, 80]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
````

````python
# Left Merge

left_merge = df1.merge(df2, on=['area', 'year'], how='left', suffixes=('_df1', '_df2'))
print(left_merge)
````

````python
# Right Merge

right_merge = df1.merge(df2, on=['area', 'year'], how='right', suffixes=('_df1', '_df2'))
print(right_merge)
````

````python
# Inner Merge

inner_merge = df1.merge(df2, on=['area', 'year'], how='inner', suffixes=('_df1', '_df2'))
print(inner_merge)
````

````python
# Outer Merge:
outer_merge = df1.merge(df2, on=['area', 'year'], how='outer', suffixes=('_df1', '_df2'))
print(outer_merge)
````

# 5. Outlier Detection and Treatment



An outlier is a data point that significantly deviates from the rest of the dataset. These anomalies can cause the distribution to become highly skewed, potentially leading to inaccurate results in later stages of analysis, such as Machine Learning. Identifying and addressing outliers is, therefore, a crucial step in the data exploration process.

**What cause an outlier?**

Outliers can typically indicate one of two scenarios:

1. Measurement Errors
- Caused by issues such as typos or malfunctions in measurement tools.
- These outliers should be carefully evaluated and, if confirmed as errors, removed from the dataset.

<br>

2. Skewed or Heavy-Tailed Distributions
- Resulting from non-normal distributions, such as exponential, geometric, Poisson, or bimodal distributions.
- In these cases, outliers may reflect the true nature of the data rather than errors.

When dealing with normally distributed data, the focus is primarily on identifying and cleaning measurement errors.

<br>

**How to detect an outlier?**

- While data visualization is a useful starting point for spotting potential outliers, mathematical methods are necessary to confirm whether a data point is truly an outlier.

1. Z-Score Method

- A data point is considered an outlier if its Z-score exceeds 3, meaning its distance from the mean is greater than three times the standard deviation.
- Formula:
<br><center>$Z score = (X - {\mu}) / {\sigma}$

  Where: X: Observed value, μ: Mean, σ:Standard deviation

<br>

2. Interquartile Range (IQR) Method

- Outliers are detected using the range between the first quartile (Q1) and the third quartile (Q3).
- Lower bound: $ Q1 - 1.5 * IQR$
- Upper bound: $ Q3 - 1.5 * IQR$
- This method is similar to the Z-score approach, as 1.5 * IQR roughly corresponds to three standard deviations.

<br>

3. Advanced Methods
For more complex datasets, advanced techniques like the [Local Outlier Factor (LOF)](https://scikit-learn.org/stable/auto_examples/neighbors/plot_lof_outlier_detection.html) or the [Minimum Covariance Determinant (MCD)](https://scikit-learn.org/stable/modules/generated/sklearn.covariance.MinCovDet.html) can be used to identify outliers more robustly.

### 5.1. Normal (Gaussian) Distribution

The [normal distribution](https://www.investopedia.com/terms/n/normaldistribution.asp#:~:text=Investopedia%20%2F%20Lara%20Antal-,What%20Is%20a%20Normal%20Distribution%3F,as%20a%20%22bell%20curve%22) is a symmetric probability distribution where most observations cluster around the mean, forming a characteristic peak at the center.
<br>
<br> Values closer to the mean are more likely to occur compared to those farther away.

Example: In a class where the mean score is 60 and the standard deviation is 20, most students' scores would fall near 60.

<br>

The normal curve, commonly referred to as the bell curve, represents the graph of the probability density function of the normal distribution. This curve shows the probabilities of various outcomes in a continuous dataset.It has two parameters:
1.	Mean (μ): Represents the center of the curve, can be any real number.
2.	Standard Deviation (σ): Indicates the spread of data around the mean, must be greater than 0.

The shape of the curve is determined by these two parameters: the mean defines the location of the peak, and the standard deviation controls the width of the curve.

This distribution is widely used in statistics and data analysis because many natural and social phenomena tend to follow a normal distribution.

**How to test for normality?**

1. **Histograms**: A normal distribution should appear as a bell-shaped curve.
2. **QQ Plot**: This plot is more effective than a histogram or box plot for detecting deviations from normality. If the data follows a normal distribution, it will align closely with the reference line, rather than forming a curve.
3. **Kolmogorov & Smirnov test**: Use this test when other techniques are inconclusive.
- If the observed data perfectly fits a normal distribution, the KS statistic will be 0.
- The test relies on the P-value to determine if the null hypothesis (that the data is normally distributed) should be rejected:
  - P-value > 0.05: Assume the data follows a normal distribution.
  - P-value ≤ 0.05: Reject the assumption of normality.
4. **Shapiro & Wilk**: Specifically designed to test for normality, unlike the KS test, which can test against other distributions.

- If the P-value of the Shapiro-Wilk Test is:
  - P-value > 0.05: Assume the data follows a normal distribution.
  - P-value ≤ 0.05: Reject the assumption of normality.
- The test provides a W statistic and a P-value.
- Note: For sample sizes larger than 5000, the W statistic remains accurate, but the P-value may not be reliable.

````python

from scipy.stats import norm
import scipy.integrate as integrate

mean = 0
distribution_function = lambda x: norm.pdf(x, mean)
x1 = -1
x2 = 1

print('probability to fall between {0} and {1} :'.format(x1, x2), integrate.quad(distribution_function, x1, x2)[0])
step = 0.001
whole_x = np.arange(-4, 4, step)
whole_y = distribution_function(whole_x)

needed_x = np.arange(x1, x2, step)
needed_y = distribution_function(needed_x)
plt.plot(whole_x, whole_y)
plt.fill_between(needed_x, needed_y, color='pink')
plt.show()

#probability to fall between -1 and 1 : 0.682689492137086
#meaning there is a 68.2% chance that the value of the variable
#will be within -1 and 1 . (-1, 1 standard deviations, first)
````

````python
# Box-cox Transformation of non-normal data to normal data

import numpy as np
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

# Generate non-normal data (exponential)
original_data = np.random.exponential(size=1000)

# Transform training data & save lambda value
fitted_data, fitted_lambda = stats.boxcox(original_data)

# Creating axes to draw plots
fig, ax = plt.subplots(1, 2)

# Plotting the original data (non-normal) and fitted data (normal)
sns.kdeplot(original_data, fill=True, linewidth=2, label="Non-Normal", color="green", ax=ax[0])
sns.kdeplot(fitted_data, fill=True, linewidth=2, label="Normal", color="green", ax=ax[1])

# Adding legends to the subplots
plt.legend(loc="upper right")

# Rescaling the subplots
fig.set_figheight(5)
fig.set_figwidth(10)

print(f"Lambda value used for Transformation: {fitted_lambda}")
````

## 6.1. Basic Plots

In data visualization, several foundational plot types help uncover patterns and trends in data. The most common and essential plot types include:

- Bar Plots: Used to compare categorical data.
- Line Plots: Ideal for showing trends or changes over time.
- Scatter Plots: Useful for analyzing relationships between two continuous variables.
- Histograms: Help visualize the distribution of a single variable.

Each plot type serves a unique purpose and is best suited for specific types of data and analyses.

### 6.1.1. Histogram Charts

Histograms are graphical representations of data frequencies, displayed as bars. They show how much of the dataset falls into specific, non-overlapping intervals called bins. The `.hist()` method in Matplotlib is used to create a simple histogram.

To plot a histogram using Matplotlib, you first need to import the library, then call the `.hist()` method on your dataset.



````python
# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np  # Missing import for numpy

# Generate random data
data1 = np.random.randn(1000)  # A sample generation based on random

# Create histogram
plt.hist(data1)
plt.xlabel("x")
plt.ylabel("Frequency")

# Show the plot
plt.show()
````

### 6.1.2. Bar charts

Bar charts represent data with rectangular bars, either vertically or horizontally. The length of each bar is proportional to the value it represents. Bar charts are ideal for comparing two or more values, and the `.bar()` method in Matplotlib can be used to create them.

````python
# Data for bar plot
data2 = [5.0, 25.0, 50.0, 20.0]

# Create bar plot
plt.bar(range(len(data2)), data2)
plt.xlabel("x")
plt.ylabel("Frequency")

# Show the plot
plt.show()
````

### 6.1.3. Scatter Plot

Scatter plots display individual data points as dots or circles, showing relationships between two continuous variables. The `.scatter()` method in Matplotlib is used to create scatter plots.

````python
import numpy as np
import matplotlib.pyplot as plt

# Generate data for plotting
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a figure with a custom size
plt.figure(figsize=(8, 6))

# Plot sine function with customized appearance
plt.plot(
    x, y, label='sin(x)', color='blue', linewidth=2, linestyle='-',
    marker='o', markersize=6, markerfacecolor='red', markeredgewidth=2,
    markeredgecolor='black'
)

# Title
plt.title('Sine Function')  # Sets the title of the plot

# Axis labels
plt.xlabel('x')
plt.ylabel('sin(x)')

# Customizing Tick Marks
plt.xticks(np.arange(0, 11, step=1))  # Set tick marks on the x-axis from 0 to 10
plt.yticks(np.arange(-1, 1.5, step=0.5))  # Set tick marks on the y-axis from -1 to 1.5

# Grid
plt.grid(True, linestyle='--', alpha=0.7)  # Add grid lines with dashed lines and transparency

# Legend
plt.legend(loc='upper right')  # Add a legend at the upper right corner

# Save the figure as a PNG file
plt.savefig('sin_plot.png')

# Display the plot
plt.show()
````

### 6.1.4. Pie Charts

Pie charts are circular graphs divided into sectors (or wedges) that represent proportions of a whole. The `.pie()` method in Matplotlib is used to create pie charts. Each wedge is proportional to the value in the dataset, calculated as x/sum(X) for each value x in the input array X.


````python
import matplotlib.pyplot as plt

# Data for pie chart
labels = 'Health', 'Climate Change', 'Obesity'
sizes = [70.4, 20.6, 9.0]
colors = ['lightgreen', 'lightcoral', 'lightskyblue']
explode = (0.1, 0, 0)

# Create pie chart
plt.pie(
    sizes, explode=explode, labels=labels, colors=colors, shadow=True,
    startangle=90
)

# Title
plt.title("What is the most concerning topic for the next 10 years? (100 people)")

# Legend
plt.legend(title="Answers")

# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')

# Display the plot
plt.show()
````

# 7. Correlation and Covariance Analysis

Correlation and covariance are important statistical tools used to measure the relationship between two or more variables.

## 7.1. Correlation Analysis and Visualization  <a class="anchor" id="four-four-one"></a>

The correlation coefficient, often denoted as "r," measures the strength and direction of the linear relationship between two variables. One of the most common methods for calculating correlation is the [Pearson correlation coefficient](https://www.statisticshowto.com/probability-and-statistics/correlation-coefficient-formula/#Pearson). The resulting value ranges from +1 to -1, indicating the strength of the relationship between two variables.

- A positive correlation (r > 0) means that as one variable increases, the other also increases at a similar rate.
- A negative correlation (r < 0) means that as one variable increases, the other decreases.
- A correlation of r = 0 indicates no linear relationship between the variables.


**What can we learn from the Correlation Heatmap?**
- In the context of regression problems, identifying variables that are highly correlated (either positively or negatively) with the target variable can be valuable. This suggests that changes in the feature will likely lead to corresponding changes in the target.

- Correlation analysis can also help identify redundant features. If two variables are highly correlated, they may provide the same information, and you could choose to remove one to reduce data complexity.

**What we should not assume from the Correlation Heatmap?**

- A correlation coefficient close to +1 or -1 indicates a strong linear relationship between two variables, but values near 0 do not necessarily mean there is no relationship. It could mean the relationship is non-linear (e.g., exponential or quadratic). Visualizing the relationship through scatter plots can help confirm this.

- It's also crucial to remember that correlation does not imply causation. A high correlation between two variables does not mean one causes the other. For example, a study might find a strong correlation between time spent watching TV and aggressiveness in children. While this could suggest that watching TV influences aggression, the reverse could be true: perhaps more aggressive children watch more TV.

**Interpreting the Correlation Heatmap**

Correlation is often visualized using a heatmap, which displays the relationships between variables. The values along the diagonal will always be 1 since a variable is always perfectly correlated with itself. For analysis, we typically focus on one half of the heatmap to examine the relationships between different variables.



````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

# Load the iris dataset as a DataFrame
iris_data = load_iris(as_frame=True)
iris = iris_data.frame

# Drop the non-numeric column ('target' is the species column in sklearn's iris dataset)
numeric_iris = iris.drop(columns=["target"])

# Calculate the correlation matrix
corr = numeric_iris.corr()

# Plot the heatmap
ax = sns.heatmap(corr, linewidths=.5, annot=True, fmt=".2f", cbar=False)
plt.show()

````

From the above plot, we can extract the following insights:
* Sepal Length is strongly correlated with Petal Length (0.87) and Petal Width (0.82), but weakly correlated with Sepal Width (-0.12).
* Sepal Width is inversely correlated with Petal Length and Petal Width; as Sepal Width increases, Petal Length and Petal Width decrease.
* Petal Length and Petal Width are highly correlated (close to 1), meaning they tend to increase together.

## 7.2. Covariance

Covariance measures how two random variables vary together. It indicates the degree to which changes in one variable can be predicted by changes in another. The sign of the covariance reveals the direction of their linear relationship:

* Positive covariance: The variables tend to move in the same direction.
* Negative covariance: The variables tend to move in opposite directions.

The example code demonstrates how to calculate covariance using the iris dataset and visualize it with a Seaborn heatmap.

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Load the iris dataset from seaborn
iris = sns.load_dataset("iris")

# Exclude the non-numeric 'species' column
numeric_iris = iris.drop(columns=["species"])

# Compute the covariance matrix
covariance_matrix = numeric_iris.cov()

# Display the covariance matrix
print(covariance_matrix)

# Create a heatmap from the covariance matrix
plt.figure(figsize=(10, 8))
sns.heatmap(covariance_matrix, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title("Covariance Matrix Heatmap")
plt.show()
````
