## 1.2. Data Quality Characteristics <a class="anchor" id="one-one"></a>

data and data collection.png

**Figure 1. Data Quality Characteristics Adapted from [here.](https://vitalflux.com/data-quality-characteristics-examples/)**

Ensuring high-quality data is crucial for effective decision-making and operations. Here are key characteristics of data quality:

- **Accuracy:** Data should precisely represent real-world facts. Inaccurate data can result from human or system errors, leading to incorrect conclusions. For instance, a person's height recorded as 15 cm is clearly incorrect.

- **Completeness:** All necessary data should be present for its intended use. Missing critical information, like a patient's allergy history, can have serious consequences. It's essential to determine which data is critical and ensure its completeness.

- **Consistency:** Data should be uniform across different systems and datasets. Inconsistencies, such as varying formats for dates or addresses, can cause confusion and errors. Regular checks are necessary to maintain consistency.

- **Timeliness:** Data must be up-to-date and available when needed. Outdated information can lead to missed opportunities or flawed strategies. For example, using old market data may result in ineffective business decisions.

- **Validity:** Data should conform to defined formats and fall within acceptable ranges. Invalid data, like a date of birth set in the future, can indicate errors in data entry or collection processes.

## 1.3. Properties of A Dataset <a class="anchor" id="one-one"></a>

A dataset is a collection of data that users can use to achieve their goals. In a dataset, rows reflect the quantity of data points, whereas columns represent the dataset's features.

**- Central Tendency:** This represents the "middle" point of the data and is typically measured using the mean, median, or mode.  It gives insight into where the majority of data points are concentrated.

**- Skewness:** This measures how balanced or symmetrical the data distribution is. A perfectly symmetrical distribution has zero skewness. Positive skewness means the data leans more towards lower values, while negative skewness indicates it leans towards higher values.

**- Data Spread:** This refers to how widely the data points are scattered around the central value. Metrics like standard deviation and variance help quantify this spread, showing how much individual data points differ from the average.

**- Outliers:** These are unusual data points that fall far outside the typical range of values.

**- Correlation:** This describes the relationship between different variables in the dataset.

**- Probability Distribution:** Understanding the data’s distribution pattern helps in predicting the likelihood of certain values occurring and selecting the right statistical tools for analysis.

## 1.6. Data Formats and File Types

In data science, working with different data formats is essential for importing, exporting, and managing data across platforms and tools. Here are some common file types and their characteristics:

**CSV (Comma Separated Values):**

- CSV files are plain text files where each line represents a row, and columns are separated by commas.
- They are simple, easy to use, and widely supported across various tools and platforms.
- Often referred to as flat files, they are commonly used for sharing and storing structured data.

**Excel (XLSX):**

- Proprietary to Microsoft, XLSX files are designed for spreadsheets and support text, numbers, formulas, and advanced formatting.
- Frequently used in business and academic settings for data storage, analysis, and reporting.

**JSON (JavaScript Object Notation):**

- A lightweight format for exchanging structured data using human-readable text.
- Popular in web applications for transmitting data between servers and clients due to its simplicity and versatility.

**Pickled Files:**

- These files store serialized Python objects, making it easy to save and load data structures for reuse.
- Useful for preserving data states or transferring Python-specific objects.

**HDF5 (Hierarchical Data Format):**

- A versatile format designed to store and organize large volumes of data efficiently.
- Widely used in data analysis for handling complex datasets, as it supports various data types and hierarchical structures.

## 2.1. For Common Files

**Importing Data:** Refers to the process of loading data from external sources (such as files, databases, or APIs) into Python programs for analysis or manipulation. This is commonly done using libraries like pandas for CSVs and Excel files, sqlite3 for databases, or requests for web-based data.

**Exporting Data:** Involves saving or writing data from Python programs to external files or databases in formats such as CSV, Excel, JSON, or SQL. This allows processed or generated data to be shared or stored for further use.

**Exporting, Importing and `open()` Function Usage for Text, CSV, JSON, Pickled, HDF5 Files**

<br>

| File Types | Exporting Data | Importing Data | `open()` Function Usage |  
|------------------|----------------|-------|------------------------------------------|  
| Text Files | df.to_csv('file.txt', sep='\t', index=False) | df = pd.read_csv('file.txt', sep='\t') | with open('file.txt', 'r') as f: <br> df = pd.read_csv(f, sep='\t') |
| CSV Files | df.to_csv('file.csv', index=False) | df = pd.read_csv('file.csv') | with open('file.csv', 'r') as f: <br> df = pd.read_csv(f) |
| JSON | df.to_json('file.json', orient='records', lines=True) | df = pd.read_json('file.json', orient='records', lines=True) | with open('file.json', 'r') as f: <br> df = pd.read_json(f, orient='records', lines=True) |

**`open()` Function Details**

The `open()` function in Python is a built-in utility used to open files and return a file object. It provides various parameters to control file access and interaction. Here's a concise explanation of its key parameters:

- **file:** Specifies the path to the file.
- **mode (default: `'r'`):** Determines how the file is opened:
  - `'r'`: Read mode.
  - `'w'`: Write mode (overwrites or creates a new file).
  - `'a'`: Append mode (adds data to the end of the file).
  - `'b'`: Binary mode.
  - `'x'`: Exclusive creation mode (fails if the file already exists).
  - `'t'`: Text mode (default).
- **buffering:** Controls buffering; -1 uses the system default.
- **encoding:** Specifies the text file encoding (e.g., `'utf-8'`).
- **errors:** Defines how encoding errors are handled (`'strict'` is the default).
- **newline:** Manages newline characters (None applies the default behavior).
- **closefd (default: `True`):** Determines whether to close the file descriptor.
- **opener:** Allows for a custom file opener, which must return an open file descriptor.

These parameters provide flexibility for working with files efficiently in Python.

## 3.2. Web Scraping  <a class="anchor" id="three-one"></a>


Web scraping is a technique used to extract data from websites, leveraging the vast array of resources available online, such as text, images, audio, and video. This data can then be processed and utilized for further computation or analysis. Web scraping is an efficient method for data collection but requires a sound understanding of tools, methodologies, and ethical considerations to use it effectively and responsibly.

**Advantages of Web Scraping**

* **Cost-Effective:** Utilizes free resources available online.
* **Automated and Fast:** Enables quick data collection with minimal manual intervention.
* **Ease of Implementation:** Accessible for users with the required technical expertise.

**Disadvantages of Web Scraping**
* **Time-Intensive Setup:** Establishing and debugging the initial codebase can take time.
* **Technical Expertise Required:** Analyzing the scraped data may pose challenges without adequate knowledge.
* **Access Restrictions:** Some websites employ protective measures or legal restrictions that complicate data extraction.

**How Web Scraping Works**

Web scraping generally involves making HTTP requests to retrieve the HTML content of web pages. This HTML is then parsed to extract the required information.

**Python and the Requests Library**
Python’s Requests library is a popular tool for web scraping. It simplifies the process of making HTTP/1.1 requests by offering:

* Sending requests like GET, POST, PUT, and DELETE.
* Handling cookies, session data, and JSON responses.
* Adding parameters and headers for enhanced functionality.
* Managing errors, redirects, and encoding seamlessly.

**Common HTTP Request Methods**
* POST (Create): Sends data to the server to create new content.
* GET (Read): Retrieves data from the server.
* PUT (Update): Updates existing data on the server.
* DELETE (Delete): Removes content from the server.

**HTTP Response Status Codes**

HTTP status codes indicate the result of a client’s request to the server:

* 2xx (Success): The request was successful, and the result is returned.
* 3xx (Redirection): Further action is required to complete the request.
* 4xx (Client Error): Indicates a bad request from the client.
* 5xx (Server Error): The server encountered an error and couldn’t process the request.

**Ethical Considerations**

When scraping websites, it is important to:

* Review and adhere to a website's terms of service.
* Avoid overloading servers with excessive requests.
* Use proper headers to simulate human-like behavior.
* Respect privacy and copyright laws while collecting data.

### 3.2.1. Importing Flat files from the Web

Flat files like CSV, TSV, or plain text files can be accessed directly from a web URL, making it convenient to work with publicly available online data without the need to manually download the files to your local machine.

In the following example, we use the `Requests` library to send an HTTP GET request to the URL hosting the data, and `Pandas` to load the CSV file directly into a DataFrame for analysis. This approach streamlines the process of working with online data sources.

````python
import pandas as pd
import io
import requests

# URL of the CSV file
url = (
    r'https://raw.githubusercontent.com/SDG-AI-Lab/Digital_Technologies_Radar/'
    'develop/src/assets/csv/technology_radar_dataset_updated_v8.csv'
)

try:
    r = requests.get(url)
    r.raise_for_status()  # Raise exception for failure

    # Read content into DataFrame
    df = pd.read_csv(io.StringIO(r.text))

except requests.RequestException as e:
    print(f"Error fetching data: {e}")

# Display first 3 rows
df.head(3)
````

**Extract Data From Webpage using Beautifulsoup**

Let's print the title of a webpage using beautifulsoup as shown below

## 3.3. Performing Requests with URL Library (urllib)

 The `urllib` library is a module in Python's standard library that provides functions and classes to work with URLs. It is used to fetch data across the web, handle URL components, encode and decode query strings, and more.

 In the example below, we show how to perform an HTTP request using Python's urllib library.



````python
# Importing necessary functions from the urllib.request module
from urllib.request import urlopen, Request

# Specifying the URL that you want to request
url = "https://sdgs.un.org/goals"
# Creating an HTTP request object for the specified URL
request = Request(url)

# Sending the request to the URL and receiving the response
response = urlopen(request)
# Reading the HTML content from the response
html = response.read()

# Printing the type of the response object to confirm the request was successful
print(type(response))

# Printing the HTML content
print(html.decode('utf-8'))  # Decode the bytes to string for better readability

# Closing the response to free up resources
response.close()
````

For such task:

2. Review the source of the page and identify what tags or attributes to extract (in this case it is "<tr>" tags and title class with specific params)
3. Store outputs in dict
4. Create dataframe
5. Save as csv

**Import libraries**

````python
# Import the 'requests' library for making HTTP requests
import requests

# Import the 'BeautifulSoup' class from the 'bs4' library for parsing HTML
from bs4 import BeautifulSoup

# Import the 'pandas' library for data manipulation and analysis
import pandas as pd
````

**Query information**

````python
months = ['01','02', '03', '04', '05', '06', '07', '08', '09', '10', '11', '12']
year = '2015'
city = 'Ankara/Central'
country = "Turkey"
````

**Table parameters inside the website table**

````python
parameters = ['Temperature avarage, Fahrenheit / Celsius', 'Temperature maximum, Fahrenheit / Celsius', 'Temperature minimum, Fahrenheit / Celsius', 'Precipitation in / cm', 'Average daily wind speed (miles/hour)',
              'Visibility in miles', 'Dew point, Fahrenheit / Celsius', 'Maximum sustained wind speed (miles/hour)', 'Maximum wind gust reported for the day', 'Sea level pressure in mBARs',
              'Station pressure in mBARs', 'Snow depth in inches', 'Weather type']
````

````python
records = {'City': [], 'Year': [], 'Month': [], 'Day': [], 'Temperature avarage, Fahrenheit / Celsius': [], 'Temperature maximum, Fahrenheit / Celsius': [], 'Temperature minimum, Fahrenheit / Celsius': [],
           'Precipitation in / cm': [], 'Average daily wind speed (miles/hour)': [], 'Visibility in miles': [], 'Dew point, Fahrenheit / Celsius': [], 'Maximum sustained wind speed (miles/hour)': [],
           'Maximum wind gust reported for the day': [], 'Sea level pressure in mBARs': [], 'Station pressure in mBARs': [], 'Snow depth in inches': [], 'Weather type': []}
````

**Table parameters inside the website table**

````python
def update_dict(records, key_name, update_element):
  # rec = records.get(key_name)
  records[key_name].append(str(update_element))
````

````python
# Iterate through months and retrieve weather data for a specific location
for month in months:
    # Construct the URL for weather data retrieval
    url = (
        "https://geographic.org/global_weather/weather_data_2.php?month={0}"
        "&year={1}&id=171300-99999&path=weather_stations/129255_171869/"
        "171300-99999.txt&name={2}&country=Turkey"
    ).format(month, year, city)

    # Make an HTTP request to the URL
    page = requests.get(url)
    text = page.text

    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(text, 'html.parser')

    # Find all table rows in the HTML content
    tr_tags = soup.find_all('tr')
    del tr_tags[0]  # Remove the header row

    check = 0
    # Iterate through table rows to extract weather information
    for table_row in tr_tags:
        # Iterate through desired parameters
        for param in parameters:
            # Find elements with the given parameter title
            row = table_row.findAll(attrs={'title': param})
            # Extract and update information in the records dictionary
            for info in row:
                update_dict(records, param, info.text)
                check = 1

        # Extract day number from the row
        day_number = table_row.find('strong').getText()

        # If parameter information was found, update additional details
        if check:
            update_dict(records, 'Day', day_number)
            update_dict(records, 'City', city)
            update_dict(records, 'Year', year)
            update_dict(records, 'Month', month)

        check = 0  # Reset the check flag
````

## 4.4  Making Query to the GitHub API:<a class="anchor" id="two-four"></a>

Let's send a query to retrieve and print some basic information about Linus Torvalds, fetched from the GitHub API.

````python
import requests

# The base URL for the GitHub API
BASE_URL = "https://api.github.com"

# Endpoint to get user information
USER_ENDPOINT = "/users/{}"

# The username you want to query information about
username = "torvalds"

# Sending GET request to the GitHub API
response = requests.get(BASE_URL + USER_ENDPOINT.format(username))

# Check if the request was successful
if response.status_code == 200:
    # Parse the JSON data
    user_data = response.json()

    # Print some information about the user
    print(f"Name: {user_data['name']}")
    print(f"Bio: {user_data['bio']}")
    print(f"Location: {user_data['location']}")
    print(f"Public Repositories: {user_data['public_repos']}")
else:
    print(f"Failed to fetch data for {username}. Status code: {response.status_code}")

````

# 5. Ethical Considerations in Data Collection



In this section, we highlight the importance of ethical considerations in data collection, focusing on legal compliance and broader ethical principles.

When dealing with sensitive data, it is crucial to understand and follow laws such as the [General Data Protection Regulation (GDPR)](https://gdpr-info.eu/), implement strong data security practices, obtain proper permissions, and ensure responsible handling of data. This approach helps protect individual privacy and upholds societal standards.

## 5.1. Data Privacy and Security


### **Understanding GDPR and Other Privacy Laws**

[General Data Protection Regulation](https://gdpr-info.eu/) (GDPR) is a European law governing personal data protection. Its key principles include consent, transparency, data minimization, and the right to be forgotten. Failure to comply with GDPR can lead to significant fines and legal penalties.

Other privacy laws, such as the California Consumer Privacy Act (CCPA), have similar provisions. It's essential to understand and comply with relevant laws in the jurisdiction where the data is collected.

**Ensuring Security in Data Collection**

* **Data Encryption:** Protect data during transmission and storage using encryption methods.
* **Access Controls:** Limit access to sensitive data by ensuring only authorized individuals can view it.
* **Regular Audits and Monitoring:** Conduct regular audits to detect and prevent unauthorized data access.
* **Incident Response Planning:** Have a plan in place to respond to data breaches quickly and minimize harm.

<br>

**Heads up! When browsing the web:**

- Avoid trusting every website or clicking on unknown links, even from trusted sources.
- Prefer secure (HTTPS) websites, especially when entering sensitive information like card numbers or IDs.
- If suspicious about a link, use a virtual keyboard when entering passwords to avoid keyloggers that may track your keystrokes.
- Avoid letting websites store your passwords, as these can be extracted using data science techniques.

## 5.2. Ethical Conduct in Data Collection


### **Gaining Proper Permissions**
* **Informed Consent:** Ensure participants understand and agree to data collection voluntarily.

* **Permission from Website Owners:** When web scraping, follow [robots.txt](https://developers.google.com/search/docs/crawling-indexing/robots/intro#:~:text=A%20robots.txt%20file%20tells,or%20password%2Dprotect%20the%20page.) files and terms of service.

* **Compliance with Institutional Guidelines:** Follow ethical review board guidelines in research.

<br>

### **Responsible Data Handling**
* **Data Minimization:** Collecting only the data that is necessary for the intended purpose.
* **Anonymization and Pseudonymization:** Using techniques to de-identify data, protecting individuals' privacy.
* **Data Retention Policies:** Defining and adhering to policies that govern how long data is retained and when it should be deleted.
* **Transparency with Stakeholders:** Clearly communicating with participants and stakeholders how data will be used, stored, and shared.
