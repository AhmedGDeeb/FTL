## 1. Exercise 1 - Python Data Structures and Loops

Create climate information for at least 5 cities using a list of dictionaries. Each city should contain:
- City name
- Temperature
- Humidity
- Rainfall

Your programme must:
1. Loop through all cities and display their information.
2. Include one city with a missing temperature value.
3. Use `continue` to skip the city with missing temperature data.
4. Classify temperatures as:
   - Extreme Heat: ≥ 40°C
   - High Heat: ≥ 35°C
   - Moderate Heat: ≥ 30°C
   - Normal: < 30°C
5. Calculate:
   - Number of valid observations
   - Average temperature
   - Highest temperature
   - Lowest temperature
6. Store the final classification results in a dictionary.

You may use built-in functions such as: `len()` • `sum()` • `min()` • `max()` • `round()`


```python
# climate information for at least 5 cities using a list of dictionaries.
# 2. Include one city with a missing temperature value: "Beirut"
cities = [
    {"name": "Cairo",     "temperature": 42, "humidity": 30, "rainfall": 2},
    {"name": "Dubai",     "temperature": 38, "humidity": 25, "rainfall": 1},
    {"name": "Riyadh",    "temperature": 35, "humidity": 20, "rainfall": 3},
    {"name": "Damascus",  "temperature": 32, "humidity": 40, "rainfall": 10},
    {"name": "Beirut",    "temperature": None, "humidity": 55, "rainfall": 20},
    {"name": "Amman",     "temperature": 28, "humidity": 45, "rainfall": 15},
]

print("=" * 50)
print("Exercise 1 - City Climate Information")
print("=" * 50)

valid_temperatures = []
classification_results = {}

# 1. Loop through all cities and display their information.
for city in cities:
    print(f"\nCity: {city['name']}")
    print(f"  Temperature: {city['temperature']}")
    print(f"  Humidity:    {city['humidity']}")
    print(f"  Rainfall:    {city['rainfall']}")

    if city["temperature"] is None:
        print("  Missing temperature data - skipping.")
        # 3. Use continue to skip the city with missing temperature data.
        continue

    temp = city["temperature"]
    valid_temperatures.append(temp)

    # 4. Classify temperatures
    if temp >= 40:
        classification = "Extreme Heat"
    elif temp >= 35:
        classification = "High Heat"
    elif temp >= 30:
        classification = "Moderate Heat"
    else:
        classification = "Normal"

    classification_results[city["name"]] = classification
    print(f"  Classification: {classification}")

# 5. Calculations
num_valid = len(valid_temperatures)
average_temp = round(sum(valid_temperatures) / num_valid, 2)
highest_temp = max(valid_temperatures)
lowest_temp = min(valid_temperatures)

print("\n" + "-" * 50)
print("Summary:")
print(f"  Valid observations: {num_valid}")
print(f"  Average temperature: {average_temp}")
print(f"  Highest temperature: {highest_temp}")
print(f"  Lowest temperature:  {lowest_temp}")

print("\nClassification Results Dictionary:")
print(classification_results)
```

## 2. Exercise 2 - Functions and Climate Risk

Create the following function:

`calculate_risk(temperature, rainfall)`

The function must return:
1. EXTREME - temperature ≥ 40°C
   HIGH - temperature ≥ 35°C or rainfall < 5mm
   MODERATE - temperature ≥ 30°C or rainfall < 15mm
   LOW - otherwise

Then:
1. Call the function for every valid city.
2. Display the city, temperature and calculated risk.
3. Create another function that returns a tuple containing: (average, minimum, maximum)
4. Display only cities classified as HIGH or EXTREME.
5. Sort the cities from the highest to the lowest temperature using `sorted()` and, where appropriate, `lambda`.


```python
print("\n" + "=" * 50)
print("Exercise 2 - Functions and Climate Risk")
print("=" * 50)

# 1. The function
def calculate_risk(temperature, rainfall):
    if temperature >= 40:
        return "EXTREME"
    elif temperature >= 35 or rainfall < 5:
        return "HIGH"
    elif temperature >= 30 or rainfall < 15:
        return "MODERATE"
    else:
        return "LOW"

# 4. The other function that returns a tuple(average, minimum, maximum)
def temperature_stats(cities):
    temps = [c["temperature"] for c in cities if c["temperature"] is not None]
    return (round(sum(temps) / len(temps), 2), min(temps), max(temps))


# 2. Call the function for every valid city.
print("\nRisk per valid city:")
for city in cities:
    if city["temperature"] is None:
        continue
    risk = calculate_risk(city["temperature"], city["rainfall"])
    print(f"  {city['name']}: {city['temperature']}C, rainfall {city['rainfall']}mm -> {risk}")

# 3. Display the city, temperature and calculated risk.
avg, minimum, maximum = temperature_stats(cities)
print(f"\nStats tuple (average, min, max): ({avg}, {minimum}, {maximum})")

# 4. Display only cities classified as HIGH or EXTREME.
print("\nCities classified as HIGH or EXTREME:")
for city in cities:
    if city["temperature"] is None:
        continue
    risk = calculate_risk(city["temperature"], city["rainfall"])
    if risk in ("HIGH", "EXTREME"):
        print(f"  {city['name']} - {risk}")

# 5. Sort the cities from the highest to the lowest temperature using sorted() and, where appropriate, lambda.
valid_cities = [c for c in cities if c["temperature"] is not None]
sorted_cities = sorted(valid_cities, key=lambda c: c["temperature"], reverse=True)

print("\nCities sorted by temperature (highest to lowest):")
for city in sorted_cities:
    print(f"  {city['name']}: {city['temperature']}C")
```

## 3. Exercise 3 - OOP Climate Monitoring System

Create a class called:

`ClimateStation`

The class must contain the following attributes:

city
temperature
humidity
rainfall

Create the following methods:

`display_summary()`
`calculate_risk()`
`update_temperature(new_temperature)`

Create at least 5 ClimateStation objects, place them inside a list and use a loop to display their information.

Then create a child class:

`SmartClimateStation`

It must inherit from ClimateStation and include:
- Additional attribute: `sensor_status`
- Method: `check_sensor()`

The child class should continue to use the methods inherited from the parent class.


```python
print("\n" + "=" * 50)
print("Exercise 3 - OOP Climate Monitoring System")
print("=" * 50)

# parent class
class ClimateStation:
    def __init__(self, city, temperature, humidity, rainfall):
        # attributes: city, temperature, humidity, rainfall
        self.city = city
        self.temperature = temperature
        self.humidity = humidity
        self.rainfall = rainfall

    # the following methods
    # --------------------------------------------------------------
    def display_summary(self):
        print(f"City: {self.city}")
        print(f"  Temperature: {self.temperature}C")
        print(f"  Humidity:    {self.humidity}%")
        print(f"  Rainfall:    {self.rainfall}mm")

    def calculate_risk(self):
        if self.temperature >= 40:
            return "EXTREME"
        elif self.temperature >= 35 or self.rainfall < 5:
            return "HIGH"
        elif self.temperature >= 30 or self.rainfall < 15:
            return "MODERATE"
        else:
            return "LOW"

    def update_temperature(self, new_temperature):
        self.temperature = new_temperature
    # --------------------------------------------------------------

# child class, inheriting ClimateStation
class SmartClimateStation(ClimateStation):
    def __init__(self, city, temperature, humidity, rainfall, sensor_status):
        super().__init__(city, temperature, humidity, rainfall)
        self.sensor_status = sensor_status

    def check_sensor(self):
        return f"Sensor status for {self.city}: {self.sensor_status}"

# testing our classes
stations = [
    ClimateStation("Cairo", 42, 30, 2),
    ClimateStation("Dubai", 38, 25, 1),
    ClimateStation("Riyadh", 35, 20, 3),
    ClimateStation("Damascus", 32, 40, 10),
    ClimateStation("Amman", 28, 45, 15),
]

print("\n--- Climate Stations ---")
for station in stations:
    station.display_summary()
    print(f"  Risk: {station.calculate_risk()}")
    print()

print("--- Updating Cairo temperature to 36C ---")
stations[0].update_temperature(36)
stations[0].display_summary()
print(f"  New risk: {stations[0].calculate_risk()}")
print()

smart_stations = [
    SmartClimateStation("Cairo", 42, 30, 2, "Active"),
    SmartClimateStation("Dubai", 38, 25, 1, "Active"),
    SmartClimateStation("Riyadh", 35, 20, 3, "Faulty"),
    SmartClimateStation("Damascus", 32, 40, 10, "Active"),
    SmartClimateStation("Amman", 28, 45, 15, "Active"),
]
```

### Bonus

Create an additional method:

`recommendation()`

that returns an action depending on the climate risk, for example:

- LOW → Normal monitoring
- MODERATE → Continue monitoring
- HIGH → Increased monitoring recommended
- EXTREME → Immediate attention required

```python
# python is "Monkey Patching" allowing dynamic method assignment / runtime method injection
# Everything in python is an object, including functions and classes
# Since classes are objects, you can define them in the runtime
def recommendation(self):
    risk = self.calculate_risk()
    if risk == "LOW":
        return "Normal monitoring"
    elif risk == "MODERATE":
        return "Continue monitoring"
    elif risk == "HIGH":
        return "Increased monitoring recommended"
    else:
        return "Immediate attention required"

SmartClimateStation.recommendation = recommendation

print("--- Smart Climate Stations ---")
for station in smart_stations:
    station.display_summary()
    print(f"  Risk: {station.calculate_risk()}")
    print(f"  {station.check_sensor()}")
    print(f"  Recommendation: {station.recommendation()}")
    print()
```
