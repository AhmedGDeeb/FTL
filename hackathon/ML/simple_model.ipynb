```python
import os
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from scipy.stats import pearsonr

from config import DATA_CLEAN, OUTPUTS
```

```python
out = os.path.join(DATA_CLEAN, "yearly_clean.csv")

df = pd.read_csv(out)
```

```python
X = df[["growing_season_rainfall_mm", "growing_season_solar_kwh_m2"]].values
y = df["wheat_yield_kg_ha"].values

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
r2   = r2_score(y, y_pred)
mae  = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print(f"Intercept: {model.intercept_:,.0f} kg/ha")
print(f"Coefficients:")
print(f"  rainfall: {model.coef_[0]:,.3f} kg/ha per mm")
print(f"  solar:    {model.coef_[1]:,.1f} kg/ha per kWh/m²/day")
print(f"MAE:  {mae:,.0f} kg/ha")
print(f"RMSE: {rmse:,.0f} kg/ha")
print(f"\nMean yield: {y.mean():,.0f} kg/ha")
print(f"RMSE as % of mean: {rmse / y.mean() * 100:.1f}%")
```

```python
out = os.path.join(OUTPUTS, "wheat_model.pkl")
joblib.dump(model, out)
```
